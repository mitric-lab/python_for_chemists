# Functions and Classes

This notebook introduces two core Python ideas:
- **functions** (reusable blocks of code), and
- **classes** (blueprints for creating objects).

We start with functions and then move to classes.

## Learning goals
By the end of this notebook, you will be able to:
- define and call functions,
- work with scopes and different argument styles,
- use `*args`, `**kwargs`, lambdas, type hints, and recursion,
- define classes and create instances,
- use constructors, attributes, and methods.

## 1) Basic function syntax

A function is defined with `def`.
It can receive inputs and (optionally) return an output with `return`.


```python
def square(number):
    result = number * number
    return result
```

```python
print(square(4))
print(square(9))
```

## 2) Local and global variables

Variables created inside a function are **local** to that function.
Variables created outside functions are **global** and can be read inside functions.


```python
def create_local_value():
    local_number = 10
    print("Inside function:", local_number)

create_local_value()
print(local_number)
```

```python
shift = 273.15

def celsius_to_kelvin(celsius):
    return celsius + shift

print(celsius_to_kelvin(25.0))
```

Best practice: prefer passing values as arguments instead of relying on globals.


## 3) Arguments

- **Positional arguments** are matched by order.
- **Keyword arguments** are matched by name.
- **Default arguments** are used if no value is provided.

```python
def kinetic_energy(velocity, mass=1.0):
    return 0.5 * mass * velocity**2
```

```python
# Positional call
print(kinetic_energy(3.0, 2.0))

# Keyword call
print(kinetic_energy(velocity=3.0, mass=2.0))

# Using default mass
print(kinetic_energy(3.0))
```

Default parameters should be placed after required parameters in function definitions.


## 4) Variable-length arguments

Use `*args` when a function should accept any number of positional arguments.
Use `**kwargs` when a function should accept any number of named arguments.


```python
def sum_all(*args):
    total = 0
    for value in args:
        total += value
    return total
```

```python
print(sum_all(1, 2, 3))

numbers = [4, 5, 6]
print(sum_all(*numbers))  # unpack list into positional arguments
```

```python
def print_person_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_person_info(name="Alice", city="Hannover")

person = {"name": "Bob", "city": "Berlin"}
print_person_info(**person)  # unpack dict into keyword arguments
```

## 5) Multiple return values

A function can return more than one value.
In practice, Python returns a tuple that you can unpack.


```python
def min_max(values):
    return min(values), max(values)
```

```python
data = [3, 8, 1, 6, 2]
smallest, largest = min_max(data)
print("min:", smallest)
print("max:", largest)
```

## 6) Lambda expressions

A lambda is a short anonymous function written in one line.
It is useful for small operations, for example sort keys.


```python
multiply = lambda x, y: x * y
print(multiply(3, 4))
```

```python
points = [(1, 3), (2, 1), (0, 2)]
points.sort(key=lambda point: point[1]) # sort by second item
print(points)
```

## 7) Type hints

Type hints describe expected argument and return types.
They improve readability and tooling, but Python does not enforce them at runtime by default.


```python
def find_index(sequence: list[int], element: int) -> int:
    for i, value in enumerate(sequence):
        if value == element:
            return i
    raise ValueError(f"{element} not found")
```

```python
print(find_index([10, 20, 30], 20))
```

## 8) Recursion

A recursive function calls itself.
Every recursive function needs a so-called **base case** to stop recursion.


```python
def factorial(n: int) -> int:
    if n == 0:
        return 1  # base case
    return n * factorial(n - 1)
```

```python
print(factorial(5))
```

Naive recursion can become slow for some problems and large inputs.


## 9) Classes

A class is a **blueprint** that describes what objects of that type look like.
An **instance** is one concrete object created from that class.


```python
class Molecule:
    pass

m1 = Molecule()
m2 = Molecule()

print(type(m1))
print(m1 is m2)
```

## 10) Constructor, attributes, and methods

- `__init__` is the constructor, called when an object is created.
- **Attributes** store data (for example `self.name`).
- **Methods** define behaviour (functions inside the class).


```python
class Particle:
    def __init__(self, name: str, mass: float, velocity: float):
        self.name = name
        self.mass = mass
        self.velocity = velocity

    def kinetic_energy(self) -> float:
        return 0.5 * self.mass * self.velocity**2

    def describe(self) -> str:
        return f"{self.name}: mass={self.mass}, velocity={self.velocity}"
```

```python
electron = Particle("electron", mass=9.11e-31, velocity=2.0e6)
print(electron.describe())
print("kinetic energy:", electron.kinetic_energy())
```
