## 1) NumPy arrays vs Python lists

A NumPy array is designed for numerical data.

At beginner level, the key differences are:
- arrays support fast elementwise math like `arr * 2 + 1`,
- arrays have a fixed shape and can be multidimensional,
- lists are general-purpose containers and do not support direct numeric vector operations.


```python
values_list = [1.0, 2.0, 3.0, 4.0]
values_array = np.array(values_list)

print("list:", values_list, type(values_list))
print("array:", values_array, type(values_array))
```

```python
print("Array arithmetic:", values_array * 2 + 1)
```

Note the fundamentally different behavior of lists and arrays:

```python
print(values_list * 2)
```

```python
print(values_list * 2 + 1)
```

Arrays make mathematical workflows cleaner because **one operation can apply to all elements at once**.


## 2) Grids with `arange` and `linspace`

In numerical programming, we often evaluate functions on a grid of x-values.
Grids are useful for tables, plots, approximations, and simulations.

- `np.arange(start, stop, step)` uses a step size.
- `np.linspace(start, stop, num)` uses a number of points.


```python
grid_a = np.arange(0.0, 1.0, 0.2)
grid_b = np.linspace(0.0, 1.0, 6)

print("arange:", grid_a)
print("linspace:", grid_b)
```

```python
y = np.sin(grid_b)
print("x-grid:", grid_b)
print("sin(x):", y)
```

Note: `linspace` includes the endpoint by default, while `arange` usually stops before `stop`.


## 3) Two- and multidimensional arrays

NumPy arrays generalize vectors and matrices to arbitrary dimensions. A 1D array is like a vector. A 2D array is like a matrix (rows and columns). A 3D array adds another axis (for example, a stack of matrices) and so on.

```python
v = np.array([1, 2, 3, 4])
M = np.array([[1, 2, 3], [4, 5, 6]])
T = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("v ndim:", v.ndim, "shape:", v.shape)
print("M ndim:", M.ndim, "shape:", M.shape)
print("T ndim:", T.ndim, "shape:", T.shape)
```

## 4) `shape` and `reshape`

The `shape` of an array tells you its size along each axis.
Use `reshape` to reorganize the same data into a different shape.
The total number of elements must stay the same.


```python
a = np.arange(12)
print("a:", a)
print("shape:", a.shape)
```

```python
A = a.reshape(3, 4)
print(A)
print("shape:", A.shape)
```

```python
B = a.reshape(2, -1)
print(B)
print("shape:", B.shape)
```

Using `-1` lets NumPy infer one dimension automatically.


## 5) Indexing NumPy arrays

Indexing lets you access one element at a specific position.

- In 1D: `x[i]`
- In 2D: `X[row, col]`
- Negative indices count from the end.


```python
x = np.array([10, 20, 30, 40])
print("x[0] =", x[0])
print("x[-1] =", x[-1])
```

```python
X = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])
print(X)
print("X[1, 2] =", X[1, 2])
print("row 0:", X[0, :])
print("column 1:", X[:, 1])
```

## 6) Slicing arrays

Slicing extracts parts of an array using:
`start:stop:step`

- `start` is included.
- `stop` is excluded.
- `step` controls spacing.

```python
s = np.arange(10)
print("s:", s)
print("s[2:7] =", s[2:7])
print("s[:5] =", s[:5])
print("s[5:] =", s[5:])
print("s[::2] =", s[::2])
print("s[::-1] =", s[::-1])
```

Let's use one fixed matrix with values `11, 12, 13, ...` so slice results are easy to read.

```python
S = np.array([
    [11, 12, 13, 14, 15],
    [21, 22, 23, 24, 25],
    [31, 32, 33, 34, 35],
    [41, 42, 43, 44, 45],
])
print(S)
```

```python
print("first two rows, first three cols:\n", S[:2, :3])
```

```python
print("rows 1..end, cols 2..end:\n", S[1:, 2:])
```

```python
print("every second row:\n", S[::2, :])
```

```python
print("every second col:\n", S[:, ::2])
```

```python
print("reverse rows and columns:\n", S[::-1, ::-1])
```

Slicing usually returns a **view** in NumPy (not a full copy). So changing the slice can also change the original array.

```python
S_view = S[:2, :2]
S_view[0, 0] = 999
print("S_view:\n", S_view)
print("S changed too:\n", S)
```

```python
S_copy = S[:2, :2].copy()
S_copy[0, 0] = -1
print("S_copy:\n", S_copy)
print("S stays unchanged by S_copy edit:\n", S)
```

## 7) Basic linear algebra with NumPy

We now use NumPy for basic linear algebra operations.


### Dot product and matrix multiplication

- `np.dot(u, v)` for dot products and matrix products.
- `@` is matrix multiplication syntax.
- `*` is elementwise multiplication.


```python
u = np.array([1.0, 2.0, 3.0])
v = np.array([4.0, 5.0, 6.0])
print("dot(u, v) =", np.dot(u, v))
```

```python
A = np.array([[1.0, 2.0], [3.0, 4.0]])
B = np.array([[2.0, 0.0], [1.0, 2.0]])

print("A @ B:\n", A @ B)
print("np.dot(A, B):\n", np.dot(A, B))
print("A * B (elementwise):\n", A * B)
```

### Norms

A norm measures vector or matrix size.
- For a 1D vector, default norm is Euclidean length.
- For a 2D matrix, default norm is Frobenius norm.


```python
w = np.array([3.0, 4.0])
print("norm(w) =", np.linalg.norm(w))

Q = np.array([[1.0, 2.0], [3.0, 4.0]])
print("Frobenius norm of Q =", np.linalg.norm(Q))
print("row norms =", np.linalg.norm(Q, axis=1))
```

### Eigenvalues and eigenvectors with `np.linalg.eigh`

Use `eigh` for symmetric matrices.
If `M` is symmetric, `eigh` returns:
- eigenvalues (1D array),
- eigenvectors (columns of a matrix).


```python
R = np.array([[2.0, -1.0], [3.0, 0.0]])
M = R + R.T  # symmetric matrix

eigvals, eigvecs = np.linalg.eigh(M)
print("M:\n", M)
print("eigenvalues:", eigvals)
print("eigenvectors (columns):\n", eigvecs)
```
