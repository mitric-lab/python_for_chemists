## 3) Operators

Operators combine or compare values.

### Arithmetic operators
Use arithmetic operators `+`, `-`, `*`, `/`, `//`, `%`, `**` to calculate new numbers from existing variables.
For example, `a + b` adds two values and returns the result, which could be stored in a new variable.

Useful notes:
- `/` gives a float result.
- `//` is floor division (integer-like quotient).
- `%` gives the remainder.
- `**` means power (for example `2**3 = 8`).


````python
a = 10
b = 3

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a**b)
````

### Relational operators

Relational operators compare two values and always return a boolean (`True` or `False`).
They are often used inside `if` statements.

Use relational operators to compare values:
- `==` checks for equality.
- `!=` checks for inequality.
- `>`, `<`, `>=`, `<=` compare numerical values.


````python
x = 7
y = 10

print("x == y:", x == y)
print("x != y:", x != y)
print("x > y:", x > y)
print("x < y:", x < y)
print("x >= 7:", x >= 7)
print("y <= 9:", y <= 9)
````

You can combine relational operators with `and`, `or`, and `not` to check multiple conditions.

````python
print("x == 7 and y == 10:", x == 7 and y == 10)
print("x == 7 or y == 1:", x == 7 or y == 1)
print("not x == 7:", not x == 7)
````

## 4) Data containers

Containers hold multiple values. Some examples are:

- **List**: ordered, mutable, initialised with square brackets `[]`
- **Tuple**: ordered, immutable, initialised with parentheses `()`
- **Dictionary**: key-value pairs, mutable, initialised with curly braces `{}`

Basic access patterns:
- List and tuple elements are accessed by index, for example `colours[0]`.
- Dictionary values are accessed by key, for example `student["name"]`.

A quick reminder:
- Indexing starts at `0`.
- Negative indices access from the end (for example `colours[-1]`).

````python
colours = ["red", "green", "blue"]
print(colours[0])

# Basic list method
colours.append("yellow")
print(colours)
````

````python
point = (2.0, 5.0)
print(point)

# Tuples are immutable, so changing an item causes an error
point[0] = 10.0
````

````python
student = {
    "name": "Alex",
    "semester": 1,
    "active": True,
}

print(student)
print("name:", student.get("name"))
print("keys:", list(student.keys()))
````

## 5) If/elif/else statements

Use `if/elif/else` to run different code depending on a condition.

How it works:
- `if` checks the first condition.
- `elif` checks additional conditions if needed.
- `else` runs when no earlier condition is `True`.

Syntax reminders:
- Each condition line ends with a colon `:`.
- The code block below it must be indented.


````python
score = 72

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"

print("Grade:", grade)
````

````python
temperature = 18
is_raining = False

if temperature > 15 and not is_raining:
    print("Good day for a walk.")
else:
    print("Better take a jacket.")
````

## 6) Loops

Loops repeat code.

- Use `for` when iterating over a sequence (like a list, tuple, string, or `range`).
- Use `while` when repeating until a condition becomes false.

Context:
- In `for item in sequence:`, `item` is the loop variable.
- `range(n)` produces numbers from `0` to `n-1`.
- As with `if` blocks, loop bodies need indentation.


````python
for colour in ["red", "green", "blue"]:
    print("Colour:", colour)
````

````python
for i in range(3):
    print("i =", i)
````

````python
counter = 0

while counter < 3:
    print("counter =", counter)
    counter += 1
````

A nested loop means one loop runs inside another.

````python
for row in range(1, 3):
    for col in range(1, 4):
        print(f"row={row}, col={col}")
````

In a `for-else` pattern, the `else` block runs if the loop finishes without `break`.


````python
numbers = [2, 4, 6, 9]

for value in numbers:
    if value % 2 != 0:
        print("Found an odd number:", value)
        break
else:
    print("All numbers are even.")
````
