# Single-layer Neural Networks



## Theoretical Foundations

$$
P(y=1|\vec{x}) = \frac{1}{1 + e^{-\vec{w}^\top \vec{x}}} = \sigma(\vec{w}^\top \vec{x})
$$

$$
\hat{y}_i 
= 
\sum_j^n a_j \sigma(\sum_k^d w_{kj} (\vec{x}_i)_k)
= 
\vec{a}^\top \sigma(\bm{W}^\top \vec{x}_i)
$$

$$
\mathcal{L} = \sum_i^N \left(\hat{y}_i - y_i \right)^2
$$

$$
\frac{\partial \mathcal{L}}{\partial \vec{a}} = \sum_i^N (\hat{y}_i - y_i) \sigma(\bm{W}^\top \vec{x}_i) 
$$

$$
\frac{\partial \mathcal{L}}{\partial \bm{W}} = \sum_i^N (\hat{y}_i - y_i) \left(\vec{a} \odot \sigma'(\bm{W}^\top \vec{x}_i) \vec{x}_i^\top \right)^\top
$$







## Implementation


```python
import numpy as np
```

```python
class Sigmoid:
    def __call__(self, x):
        return 1 / (1 + np.exp(-x))

    def gradient(self, x):
        return self(x) * (1 - self(x))
```

```python
class SLP:
    def __init__(self, dim=2, hidden_size=2, activation='Sigmoid', n_epochs=100, alpha=0.1, batch_size=5):
        self.weights = np.random.randn(dim + 1, hidden_size)
        self.linear_weights = np.random.randn(hidden_size)
        if activation == "Sigmoid":
            self.activation = Sigmoid()
        else:
            raise NotImplementedError(f"Activation function not implemented.")
        self.epochs = n_epochs
        self.alpha = alpha
        self.batch_size = batch_size
    
    def feedforward(self, x): 
        z = np.dot(self.weights.T, x) + self.bias
        return np.dot(self.linear_weights.T, self.activation(z))
```

```python
    def train(self, X, y):        
        N = X.shape[0]
        for e in range(self.epochs):
            # Iterate over batches
            for i in range(0, N, self.batch_size):

                # Define batch
                X_batch = X[i:i + self.batch_size]
                y_batch = y[i:i + self.batch_size]

                # Initialize gradients
                grad_w = np.zeros_like(self.weights)
                grad_lw = np.zeros_like(self.linear_weights)

                # Accumulate gradients over the batch
                for xi, yi in zip(X_batch, y_batch):
            
                    zi = np.dot(self.weights.T, xi)
                    d_inner = self.linear_weights * self.activation.gradient(zi)
                    residue = self.feedforward(xi) - yi

                    # Compute gradients
                    grad_w += residue * np.outer(d_inner, xi).T
                    grad_lw += residue * self.activation(zi)

                # Update parameters after each batch
                self.weights -= self.alpha / self.batch_size * grad_w
                self.linear_weights -= self.alpha / self.batch_size * grad_lw
```
