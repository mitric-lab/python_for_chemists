# Multi-layer Neural Networks

- what we have seen so far is a single layer neural network, or more precisely, a neural network with a single *hidden* layer
- already demonstrated in the previous chapter that such a model is capable of learning complex patterns, such as non-linear decision boundaries
- in fact, it can be mathematically shown that single-layer neural networks are capable of *representing any continuous function to arbitrary accuracy*, given enough neurons
- this is known as the **universal approximation theorem**
- in practice, however, it has been found that scaling neural networks by stacking multiple hidden layers is much more effective, leading to mulit-layer or **deep neural networks**

## Theoretical Foundations

### Fully-Connected Neural Networks

- to get a detailed understanding of the mathematical properties of deep neural networks, let us first consider a single layer neural network again
- {ref}`fig:single_layer_perceptron` illustrates the computation of the network output $\hat{y}$ (red), where compared to {ref}`fig:neuron`, we have combined the summation of weighted input features (blue) and activation function into a single layer of nodes (green)

:::{figure} /assets/figures/deep_learning/slp.svg
:label: fig:single_layer_perceptron
:align: center
:width: 300px
:alt: Computational graph of logistic regression
A computational graph of logistic regression.
:::

- formally, the output of this single layer with $n$ hidden neurons can be written as a vector

$$
\vec{h} = \sigma(\bm{W}^\top \vec{x}) \in \mathbb{R}^n
$$

- from this abstract representation, the generalisation to a multi-layer network is straightforward: we simply add additional layers of nodes between the input and output, as illustrated in {ref}`fig:multi_layer_perceptron`

:::{figure} /assets/figures/deep_learning/mlp.svg
:label: fig:multi_layer_perceptron
:align: center
:width: 500px
:alt: Computational graph of logistic regression
A computational graph of logistic regression.
:::

- thus, each layer takes as input the output of the previous layer:

$$
\vec{h}_l = \sigma_l(\bm{W}_l^\top \vec{h}_{l-1}) \in \mathbb{R}^{n_l},
$$


- where we have again assumed
- crucially, each layer can have its own number of neurons $n_l$, activation function $\sigma_l$ and has independent weights $\bm{W}_l \in \mathbb{R}^{n_l \times n_{l-1}}$ that **fully connect** it with the previous layer
- for a given feature vector $\vec{x}_i$, the output $\hat{\vec{y}}_i$ of the network is then given by the **composition** of all layers:

$$
\hat{\vec{y}}_i = \vec{h}_L \circ \vec{h}_{L-1} \circ \cdots \circ \vec{h}_2 \circ \vec{h}_1(\vec{x}_i)
$$

- here, we have assumed the output $\hat{\vec{y}}_i$ to be vector, which at the same time constitutes the last layer $\vec{h}_L$ of the network

### Backpropagation

- training any deep neural network, i.e. finding the set of weights $\bm{W}_l$ for all layers $l = 1, \ldots, L$ that minimise a given loss function $\mathcal{L}$ over a training data set follows known techniques, such as mini-batch gradient descent
- requires the computation of the gradient of the loss function with respect to the weights of each layer
- because of the layered structure of the network, this gradient can be derived by repeated application of the chain rule
- as a result, gradients with respect to weights in any layer require the gradients of all subsequent layers
- which is why we start by computing the gradients from last to first layer, algorithmically known as **backpropagation**
