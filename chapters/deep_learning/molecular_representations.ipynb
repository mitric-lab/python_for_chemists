# Molecular Representations

- In this section, present practical considerations for applying deep learning in chemistry
- while there are countless use cases of deep learning in chemistry, we will focus on problems that involve molecular data, e.g. predicting properties of molecular species
- here, we will discuss how to represent molecules in a way that can be used by machine learning models
- has been shown that the choice of representation can have a large impact, comparable to the impact of the choice of model architecture
