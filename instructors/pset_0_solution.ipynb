## Exercise 0.2: Functions and Classes

**(a)** Write a function `cube(x)` that returns `x**3`.

```python
# SOLUTION-START
def cube(x):
    return x**3
# SOLUTION-END
```

**(b)** Write a function with one required argument and one default argument.

```python
# SOLUTION-START
def molarity(moles, volume_L=1.0):
    return moles / volume_L

print(molarity(0.5), molarity(0.5, volume_L=2.0))
# SOLUTION-END
```

**(c)** Write a function that returns two values and unpack them.

```python
# SOLUTION-START
def mass_fractions(total_g, water_mass_frac):
    water_g = total_g * water_mass_frac
    solid_g = total_g - water_g
    return water_g, solid_g

w, s = mass_fractions(100.0, 0.85)
print(w, s)
# SOLUTION-END
```

**(d)** Use `lambda` as a `key` to sort a list of tuples by the first item.

```python
# SOLUTION-START
elements = [(10, "Ne"), (2, "He"), (18, "Ar")]
elements.sort(key=lambda t: t[0])
print(elements)
# SOLUTION-END
```

**(e)** Create a class with two attributes and one method, then create one instance and call the method.

```python
# SOLUTION-START
class Molecule:
    def __init__(self, formula, molar_mass):
        self.formula = formula
        self.molar_mass = molar_mass

    def moles(self, mass_g):
        return mass_g / self.molar_mass


water = Molecule("H2O", 18.015)
print(water.moles(36.03))
# SOLUTION-END
```

**(c)** Add `xlabel`, `ylabel`, `title`, `xlim`, and `ylim` to one plot.

```python
# SOLUTION-START
x = np.linspace(0, 2 * np.pi, 100)
plt.plot(x, np.sin(x))
plt.xlabel("phase (rad)")
plt.ylabel("signal (arb.)")
plt.title("Detector oscillation")
plt.xlim(0, np.pi)
plt.ylim(0, 1.05)
plt.show()
# SOLUTION-END
```

**(d)** Plot `sin(x)` and `cos(x)` in one figure with different styles and a legend.

```python
# SOLUTION-START
x = np.linspace(0, 2 * np.pi, 50)
plt.figure()
plt.plot(x, np.sin(x), "--", label="sin")
plt.plot(x, np.cos(x), "-.", label="cos")
plt.legend()
plt.show()
# SOLUTION-END
```

**(e)** Build a 2x2 subplot figure and place one function in each panel.

```python
# SOLUTION-START
x = np.linspace(0, 2 * np.pi, 50)
fig, axes = plt.subplots(2, 2, figsize=(6, 5))
axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, np.sin(x) ** 2)
axes[1, 1].plot(x, np.exp(-x / np.pi))
plt.tight_layout()
plt.show()
# SOLUTION-END
```
