## Aufgabe 2: Gradientenverfahren

Ein Student soll das Gradientenverfahren implementieren und damit ein **lokales** Minimum eines Doppelmuldenpotentials finden. Ihm wurde die Iterationsgleichung des Gradientenverfahrens

$$
x_{n+1} = x_n - \tau \nabla f(x_n)
$$

ausgehend vom Startpunkt $x_0$ mit der Schrittlänge $\tau$, sowie die Definition des zu untersuchenden Doppelmuldenpotentials

$$
f(x) = x^6 - 2x^2 - \frac{1}{4}x
$$

gegeben. Da er aber die Vorlesung nur selten besuchte, ist der Code fehlerhaft.

Finden und korrigieren Sie alle **5 fehlerhaften Zeilen** (sowohl Syntaxfehler als auch inhaltliche Fehler) im folgenden Code-Block.

**Lösung:** Die fünf fehlerhaften Zeilen sind:

1. **Gradient:** Die Ableitung von $-2x^2$ ist $-4x$, nicht $-2x$.
2. **Funktionskopf:** Nach der Parameterliste fehlt der Doppelpunkt.
3. **Initialisierung:** Der Startwert muss `x = x0` sein, nicht `x = 0`.
4. **Abbruchkriterium:** Der Gradient kann negativ sein; daher muss `abs(grad) < maxgrad` geprüft werden.
5. **Rückgabe:** `return x, converged` darf nicht innerhalb der `for`-Schleife stehen.

Korrigierter Code:

````python
def double_well_gradient(x):  # x: float
    grad = 6 * x**5 - 4 * x - 0.25
    return grad

def gradient_descent(func_grad, x0, tau=0.01, maxgrad=1e-6, maxiter=500):
    x = x0
    converged = False

    for _ in range(0, maxiter):
        grad = func_grad(x)
        x = x - tau * grad
        if abs(grad) < maxgrad:
            converged = True
            break

    return x, converged

x_opt, converged = gradient_descent(double_well_gradient, -1.2)
print("Ein lokales Minimum liegt bei x =", x_opt)
print("Konvergenz erreicht:", converged)
````

**Lösung (b), Variante 1:** Zuerst alle Distanzen berechnen, dann die Indizes der $k$ nächsten Nachbarn sortieren und schließlich die Mehrheitsklasse mit einfachen Zählungen bestimmen.

````python
class kNNClassifier:
    def __init__(self, k):
        self.k = k

    def predict(self, X, y, xi):
        distances = np.sqrt(np.sum((X - xi) ** 2, axis=1))
        sorted_indices = np.argsort(distances)
        nearest_labels = y[sorted_indices[:self.k]]

        count_0 = np.sum(nearest_labels == 0)
        count_1 = np.sum(nearest_labels == 1)

        if count_0 >= count_1:
            y_pred = 0
        else:
            y_pred = 1

        return y_pred
````

Das Modell soll wie folgt verwendet werden:

````python
np.random.seed(0)
N = 20
X = np.random.randn(N, 2)
y = np.hstack((np.zeros(N//2), np.ones(N//2)))

knn = kNNClassifier(k=3)

xi = np.array([0, 0])
y_pred = knn.predict(X, y, xi)

print(y_pred)  # Output: 0 or 1
````
